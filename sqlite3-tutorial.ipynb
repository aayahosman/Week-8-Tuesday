**Some definitions before we start**

**📌  What is SQLite?**  
SQLite is a lightweight, file-based relational database engine.  
It does not require a separate server and stores data in a single file.  
It is widely used in mobile apps, browsers (like Chrome), and embedded systems.  

**💡 Analogy:**
Think of SQLite as a notebook where you write down important data.

You can write, read, and erase information in the notebook without needing a complex storage system.

**📌 What is `sqlite3`?**  
sqlite3 is the Python standard library module that allows Python programs to interact with an SQLite database.  
It provides functions to create tables, insert data, execute queries, and manage transactions.

✅ Why sqlite3 and not just sqlite?

The name `sqlite3` reflects that it is an interface for SQLite version 3, which is the modern version of SQLite.  
Older versions (like SQLite2) had a different API, and sqlite3 was introduced to support SQLite3 syntax.

### 📌 3. Create a Table

````python
cursor.execute("""
CREATE TABLE IF NOT EXISTS users (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER,
    city TEXT
)
""")
conn.commit()  # Save changes
````

✅ Use AUTOINCREMENT for automatic ID assignment.  
✅ Use NOT NULL to ensure a column cannot have empty values.

### Aside: Explain primary key with analogy 

A **Primary Key** in a database is a unique identifier for each record in a table, ensuring that each row is distinct and easily referenced.

**🎟️ Analogy: A College Student Pass**
Recall that every student at a college receives a student pass (ID card) when they enroll.  

* Each student gets a unique student ID number printed on the card.  
* No two students can have the same student ID number (it must be unique).  
* If a student changes their name, major, or dormitory, their student ID number remains the same.  
* If the student ID number is missing, the system cannot identify the student.  

**💡 Database Parallel:**

* The Primary Key is like the student ID number—it uniquely identifies each record.  
* Each row in a database table represents a unique entity, just as each student is an individual in the college system.  
* The Primary Key ensures that no two records share the same identifier, just like no two students can have the same student ID.  
* Even if other attributes in a row change (like a student’s name or major), the Primary Key remains constant, maintaining a stable reference for each record.

### Aside: Explain SQL injection attacks with an analogy

An SQL Injection Attack happens when a hacker manipulates an input field to execute unauthorized database commands, potentially stealing, modifying, or deleting data.

**🍕 Analogy: Ordering Pizza Over the Phone** . 
Imagine you’re calling a pizza restaurant to place an order. The conversation goes like this:  

📞 You: "I'd like to order one large pizza."  
👨‍🍳 Employee: "Got it! One large pizza." . 

✅ Everything works as expected—you placed a valid order.  

**Now, what if someone tries to trick the system?**  
📞 Hacker: "I'd like to order one large pizza AND cancel all other orders."  
👨‍🍳 Employee (Without Verification): "Okay, canceling all orders!"  

❌ The restaurant was tricked into performing an unintended action, just because they didn’t verify the input properly.


**💡 How This Relates to SQL Injection** 
The pizza order is like an SQL query.  
The restaurant employee is like the database—they execute whatever is requested without checking.  
The hacker’s trick is like an SQL injection attack, where a malicious user inserts extra commands into an input field.  

**📌 Example in SQL**
Let’s say we have a login system where a user enters their username and password.

❌ Vulnerable Code (Unsafe Query)

```
username = input("Enter username: ")
password = input("Enter password: ")

query = f"SELECT * FROM users WHERE username = '{username}' AND password = '{password}'"
cursor.execute(query)  # 🚨 Vulnerable to SQL Injection!
```

If a hacker enters this in the username field:  

`' OR 1=1 --`

The actual SQL query becomes:

`SELECT * FROM users WHERE username = '' OR 1=1 -- AND password = ''`

* `OR 1=1` always evaluates to true, allowing the hacker to log in without a password.
* -- comments out the rest of the query, preventing errors.

**✅ How to Prevent SQL Injection**

Use prepared statements with parameterized queries: 

`cursor.execute("SELECT * FROM users WHERE username = ? AND password = ?", (username, password))`

✔ This treats inputs as data, not executable SQL code.  
✔ Even if a hacker enters `' OR 1=1 --`, it won’t work.

### 📌 6. Fetch Data  
🔹 Fetch All Rows

````python
cursor.execute("SELECT * FROM users")
rows = cursor.fetchall()  # Gets all rows
for row in rows:
    print(row)
````

🔹 Fetch Only One Row

````python
cursor.execute("SELECT * FROM users WHERE name = ?", ("Alice",))
row = cursor.fetchone()
print(row)
````

✅ fetchone() gets only one result.  
✅ fetchall() returns all matching rows.

### Aside: Why use tuple values such as ("Alice",) in `cursor.execute`?

When using parameterized queries in SQLite (or any SQL execution in Python), values must be passed as a tuple. The reason is how SQLite expects the data to be formatted when passing parameters.  

**✅ Why Use a Tuple?**

When writing: `cursor.execute("SELECT * FROM users WHERE name = ?", ("Alice",))`, the second argument is ("Alice",), which is a tuple with one element.

**⚠️ What Happens If You Don't Use a Tuple?**

If you write: `cursor.execute("SELECT * FROM users WHERE name = ?", "Alice")  # ❌ WRONG!`, it will raise an error because "Alice" is treated as a string, not a sequence of values.

**📌 Why is the Comma Important?**

In Python, a single value inside parentheses is not a tuple:
```
x = ("Alice")  # This is just a string!
y = ("Alice",)  # ✅ This is a tuple!
print(type(x))  # ❌ <class 'str'>
print(type(y))  # ✅ <class 'tuple'>
```

To ensure SQLite recognizes the parameter correctly, you must include a trailing comma when passing a single value.

**✅ Summary**   
✔ Always use a tuple when passing parameters to execute().  
✔ For a single value, include a comma (("Alice",)).  
✔ For multiple values, use a tuple naturally (("Alice", 25)).

### 📌 10. Use `with ... as` statement (Best Practice)

Using `with ... as` ensures the connection is properly closed:

````python
with sqlite3.connect("my_database.db") as conn:
    cursor = conn.cursor()
    cursor.execute("SELECT * FROM users")
    rows = cursor.fetchall()
````

✅ Auto-closes connection when with block ends.

### 📌 12. Check if a Table Exists in the database

````python
cursor.execute("SELECT name FROM sqlite_master WHERE type='table' AND name='users'")
exists = cursor.fetchone()
if exists:
    print("Table exists!")
````

✅ Use sqlite_master to check table existence.

### Aside: Explain sqlite_master with an analogy

In SQLite, sqlite_master is a special system table that keeps track of all other tables, indexes, views, and triggers in the database.  

**🏛️ Analogy: A Library Catalog**  
Imagine a library 📚 that has hundreds of books. To keep track of them, the library maintains a catalog that records:

* The title of each book 📖 (like table names in a database).  
* The author and subject (like columns in a table).  
* Whether the book is fiction, non-fiction, or reference (like table types: regular table, view, index, etc.).  

Whenever you need to find a book, you don’t scan every shelf manually—instead, you check the library catalog first.  

💡 In the same way, SQLite doesn’t search all tables directly—it first looks at sqlite_master to find where everything is stored.

````python
cursor.execute("SELECT name, type FROM sqlite_master WHERE type='table'")
tables = cursor.fetchall()
tables
````

### 📌 13. Use pandas with sqlite3

**Store a database table into a pandas dataframe**

````python
import pandas as pd

conn = sqlite3.connect("my_database.db")
df = pd.read_sql_query("SELECT * FROM users", conn)
print(df)
````

✅ Reads data directly into a pandas DataFrame.

**Store a dataframe as a table into the database**

Since we don't have a dataframe, we will create it first, but one can read the dataframe from a CSV file as well.

````python
# Create a sample DataFrame for courses
courses_data = {
    "course_name": ["Math 101", "History 202", "Physics 303"],
    "instructor": ["Dr. Smith", "Prof. Johnson", "Dr. Lee"],
    "credits": [3, 4, 3]
}

df_courses = pd.DataFrame(courses_data)

# Insert DataFrame into SQLite as a new table "courses"
df_courses.to_sql("courses", conn, if_exists="replace", index=False)

# Test that it was added in the database
cursor.execute("SELECT * FROM courses")
rows = cursor.fetchall()  # Gets all rows
for row in rows:
    print(row)
````

## My Netflix K-drama Database

````python
import sqlite3
import pandas as pd

netflix_conn = sqlite3.connect("simple_netflix.db")
netflix_cursor = netflix_conn.cursor()
````

````python
netflix_df = pd.read_csv("netflix_top10_countries.tsv", sep="\t")

dramas = ["True Beauty", "Mask Girl", "My ID is Gangnam Beauty"]

small_df = netflix_df[
    (netflix_df["show_title"].isin(dramas)) &
    (netflix_df["category"] == "TV")
].copy()

small_df.head()
````

````python

netflix_cursor.execute("""
CREATE TABLE IF NOT EXISTS dramas (
    country_name TEXT,
    week TEXT,
    show_title TEXT,
    weekly_rank INTEGER,
    score INTEGER
)
""")

netflix_conn.commit()
````

````python

def rank2score(rank):
    points = [25, 18, 15, 12, 10, 8, 6, 4, 2, 1]
    return points[rank - 1]

small_df["score"] = small_df["weekly_rank"].apply(rank2score)
````

````python
netflix_cursor.execute("DELETE FROM dramas")

for _, row in small_df.iterrows():
    netflix_cursor.execute(
        "INSERT INTO dramas VALUES (?, ?, ?, ?, ?)",
        (
            row["country_name"],
            row["week"],
            row["show_title"],
            int(row["weekly_rank"]),
            int(row["score"])
        )
    )

netflix_conn.commit()
````

````python
# See the first five records
netflix_cursor.execute("SELECT * FROM dramas LIMIT 5")
netflix_cursor.fetchall()
````

````python

# See the True Beauty records in South Korea
netflix_cursor.execute("""
SELECT *
FROM dramas
WHERE show_title = 'True Beauty'
AND country_name = 'South Korea'
""")

netflix_cursor.fetchall()
````

````python

# Find the highest scoring countries for Mask Girl
netflix_cursor.execute("""
SELECT country_name, SUM(score) AS total_score
FROM dramas
WHERE show_title = 'Mask Girl'
GROUP BY country_name
ORDER BY total_score DESC
LIMIT 10
""")

netflix_cursor.fetchall()
````

````python
netflix_conn.close()
````
