### Q8: Time frame for a given country

Write a function that given a country name prints out the start and end date of their rankings.

Once you have created the function, call it for a few countries to test it.

```python

def show_timeframe(con, country):
    """A function to print out the time frame a country was in the Netflix rankings using DuckDB."""
    res = con.execute("""
        SELECT MIN(week), MAX(week)
        FROM (
            SELECT country_name, week FROM tvseries
        )
        WHERE country_name = $1
    """, [country]).fetchone()

    start, end = res[0], res[1]
    print(f"Netflix rankings for {country} exist between {start} and {end}.")

show_timeframe(con, "Chile")
show_timeframe(con, "Russia")
```

Check if total rows match `20 * unique_weeks * unique_countries`.

<a id="q10"></a>
### Q10: Country with fewest weeks

Find the country with the fewest weeks in the Netflix rankings.

```python
con.execute("""
SELECT country_name, COUNT(DISTINCT week) AS weekCnt
FROM tvseries
GROUP BY country_name
ORDER BY weekCnt ASC
LIMIT 5;""")
```

## Netflix F1 Ranking Analysis with DuckDB

```python

import duckdb

con = duckdb.connect("netflix_data.db")

con.execute("""
CREATE OR REPLACE TEMP VIEW netflix_updated AS
SELECT *
FROM read_csv('netflix_top10_countries.tsv', delim='\t', header=true)
""")
```

```python

con.execute("""
SELECT
    show_title,
    country_name,
    weekly_rank,
    CASE weekly_rank
        WHEN 1 THEN 25
        WHEN 2 THEN 18
        WHEN 3 THEN 15
        WHEN 4 THEN 12
        WHEN 5 THEN 10
        WHEN 6 THEN 8
        WHEN 7 THEN 6
        WHEN 8 THEN 4
        WHEN 9 THEN 2
        WHEN 10 THEN 1
    END AS score
FROM netflix_updated
LIMIT 10
""")
```

```python

con.execute("""
SELECT
    country_name,
    SUM(
        CASE weekly_rank
            WHEN 1 THEN 25
            WHEN 2 THEN 18
            WHEN 3 THEN 15
            WHEN 4 THEN 12
            WHEN 5 THEN 10
            WHEN 6 THEN 8
            WHEN 7 THEN 6
            WHEN 8 THEN 4
            WHEN 9 THEN 2
            WHEN 10 THEN 1
        END
    ) AS total_score,
    COUNT(DISTINCT week) AS weeks
FROM netflix_updated
WHERE show_title = 'True Beauty'
AND category = 'TV'
GROUP BY country_name
ORDER BY total_score DESC
LIMIT 10
""")
```
